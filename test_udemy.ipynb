```python
from selenium import webdriver
from selenium.common.exceptions import TimeoutException
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait

url = "https://www.udemy.com/course/complete-microsoft-sql-masterclass-beginner-advanced/"

with webdriver.Chrome() as driver:
    driver.get(url)

    try:
        heading = WebDriverWait(driver, 15).until(
            EC.visibility_of_element_located(
                (By.XPATH, "//h1[contains(., 'Microsoft SQL Server Masterclass')]")
            )
        )
        print("Course heading:", heading.text)
    except TimeoutException:
        print("Final URL:", driver.current_url)
        print("Page title:", driver.title)
        print("H1 headings:", [
            h.text for h in driver.find_elements(By.TAG_NAME, "h1")
            if h.is_displayed()
        ])
        print("Page text:", driver.find_element(By.TAG_NAME, "body").text[:1200])
```

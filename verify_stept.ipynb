```python
from selenium import webdriver
from selenium.common.exceptions import InvalidArgumentException
from selenium.webdriver.firefox.options import Options

options = Options()
options.add_argument("--headless")

try:
    with webdriver.Firefox(options=options) as driver:
        driver.get("https://www.python.org/")
        print(driver.title)
except InvalidArgumentException as exc:
    print(exc.msg)
```

# Extracting title from firefox

```python
from selenium import webdriver
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.firefox.service import Service

options = Options()
options.add_argument("--headless")
options.binary_location = "/snap/firefox/current/usr/lib/firefox/firefox"

service = Service(executable_path="/snap/bin/geckodriver")

with webdriver.Firefox(options=options, service=service) as driver:
    driver.get("https://www.python.org/")
    print(driver.title)
```

```python
driver.quit()
```

```python
import os

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By

chrome_env = os.environ.copy()
chrome_env.pop("GSETTINGS_SCHEMA_DIR", None)
chrome_env["XDG_DATA_DIRS"] = "/usr/local/share:/usr/share"

# Replace this with the results page that contains your button and track cards.
url = "https://bandcamp.com/discover//"

with webdriver.Chrome(service=Service(env=chrome_env)) as driver:
    driver.get(url)
    print("Page title:", driver.title)

    pagination_buttons = driver.find_elements(By.ID, "view-more")
    if pagination_buttons:
        print("Button accessible name:", pagination_buttons[0].accessible_name)
    else:
        print("No element with ID 'view-more' was found.")

    tracks = driver.find_elements(By.CLASS_NAME, "results-grid-item")
    print("Number of tracks:", len(tracks))

    if tracks:
        print("First track:", tracks[0].text)
```

# Pagination for firefox

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.firefox.service import Service

options = Options()
options.binary_location = "/snap/firefox/current/usr/lib/firefox/firefox"

service = Service(executable_path="/snap/bin/geckodriver")

# Replace this with the URL of your actual results page.
url = "https://www.microsoft.com/sl-si/"

with webdriver.Firefox(options=options, service=service) as driver:
    driver.get(url)
    print("Page title:", driver.title)

    pagination_buttons = driver.find_elements(By.ID, "view-more")
    if pagination_buttons:
        print("Button accessible name:", pagination_buttons[0].accessible_name)
    else:
        print("No element with ID 'view-more' was found.")

    tracks = driver.find_elements(By.CLASS_NAME, "results-grid-item")
    print("Number of tracks:", len(tracks))

    if tracks:
        print("First track:", tracks[0].text)
```
